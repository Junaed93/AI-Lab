How ```enumerate()``` Works Internally
```
def my_enumerate(iterable, start=0):
    index = start
    for item in iterable:
        yield index, item
        index += 1
```

```
for i, fruit in my_enumerate(fruits):
    print(i, fruit)
```


## Slicing
Slicing is a powerful feature in Python that allows you to extract a portion of a sequence (such as lists, tuples, strings, and more.
The slicing syntax follows this structure:
```
sequence[start:stop:step]
```

*   start (optional): The index where the slice starts (inclusive). Default is 0.
*   stop (optional): The index where the slice stops (exclusive). If omitted, it goes to the end.
* step (optional): The step size (or stride). Default is 1

Write a code that counts the frequency of each letter in a sentence

```
sentence = "I study in UIU"

freq = {}
```

````python
sentence = "I study in UIU"

freq = {}
````

````python
for i in range(len(sentence)):
    # print(sentence[i])
    char = sentence[i]
    if char not in freq:
        freq[char] = 0
    freq[char] = freq[char]+1

for char in freq:
    print(f"{char}: {freq[char]}")
````

````python
sorted_freq = dict(sorted(freq.items()))

for char, count in sorted_freq.items():
    print(f"{char}: {count}")
````

```*args (Non-keyword arguments)```

How It Works?
- ```*args``` collects all positional arguments as a tuple.
- It allows calling the function with different numbers of arguments.

````python
def add_numbers(*args):
    return sum(args)

print(add_numbers(2, 3))
print(add_numbers(1, 2, 3, 4))
````

We can also use *args to unpack a list or tuple when calling a function.

````python
numbers = [1, 2, 3, 4]
print(add_numbers(*numbers))
````

```**kwargs``` (Keyword arguments)

- ```**kwargs``` allows a function to accept any number of keyword arguments.

- The arguments are passed as a dictionary inside the function.

````python
def print_details(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}: {value}")
````

````python
print_details(name="SHM", age=25, city="Dhaka")
````

We can also use **kwargs to unpack a dictionary when calling a function.


````python
info = {'name': 'Ratul', 'age': 30}
print_details(**info)
````

````python
def full_function(a, b, *args, **kwargs):
    print(f"a: {a}, b: {b}")
    print("args:", args)
    print("kwargs:", kwargs)

full_function(1, 2, 3, 4, 5, name="Messi", age=36)
````

## File

Opening a File
```
file = open("filename.txt", mode)
```

- "r"	Read mode (default). Raises an error if the file doesn’t exist.
- "w"	Write mode. Creates a new file or overwrites an existing one.
- "a"	Append mode. Adds data to the end of the file without deleting existing content.
- "x"	Exclusive creation. Fails if the file already exists.
- "b"	Binary mode (used with "rb", "wb", etc.) for non-text files like images.
- "t"	Text mode (default). Handles text files.
- "+"	Read and write mode (used with "r+", "w+", "a+").

````python
def find_longest_word(filename):
    with open(filename, "r") as file:
      read = file.read()
      words = read.split()
      longest_word = max(words, key=len)
      print("Longest word:", longest_word)

find_longest_word("text.txt")
````

````python
def search_word(filename, word):
    with open(filename, "r") as file:
        lines = file.readlines()

    for i, line in enumerate(lines, start=1):
        if word in line:
            print(f"Found '{word}' on line {i}")

search_word("text.txt", "mango")
````

````python
import string

def generate_alphabet_dict():
    alphabet_dict = {}
    for i, char in enumerate(string.ascii_lowercase, 1):
        alphabet_dict[i] = char
    return alphabet_dict

alphabet_dict = generate_alphabet_dict()
````

````python
def decode_message(n):
    stri = ""
    for i in n.split(" "):
        stri += alphabet_dict[int(i)]
    return stri
````

````python
print(decode_message("8 5 12 12 15"))
````

````python
def write_students(filename):
    with open(filename, "w") as file:
        while True:
            name = input("Enter student name (or 'exit' to stop): ")
            if name.lower() == "exit":
                break
            marks = input("Enter marks: ")
            file.write(f"{name} - {marks}\n")

def read_students(filename):
    with open(filename, "r") as file:
        print("\nStudent Records:")
        print(file.read())

filename = "students.txt"
write_students(filename)
read_students(filename)
````
